```python
#두개의 숫자를 입력 받아서 더한 값을 반환하는 함수

#함수 이름 : first_sum

def first_sum(score1,score2):
    result = score1 + score2
    return result
num1 = int(input("첫번째 숫자: "))
num2 = int(input("두번째 숫자: "))

ret = first_sum (num1,num2)

print("결과 : ", ret)
    
```
