```python

try:
    age = int(input("나이를 입력하세요: "))
    print(f"입력한 나이: {age}")
except ValueError:
    print("나이는 숫자로 입력해 주세요.")
```

```python

try:
    number = int(input("숫자를 입력하세요: "))
    result = 100 / number
    print(result)
except ValueError:
    print("정수를 입력해 주세요.")
except ZeroDivisionError:
    print("0으로 나눌 수 없습니다.")
```

```python

try:
    number = int(input("정수: "))
except ValueError:
    print("정수가 아닙니다.")
else:
    print(f"입력 성공: {number}")
finally:
    print("입력 처리를 마쳤습니다.")
```

```python
try:
    number = int(input("정수: "))
    print(f"입력 성공: {number}")
except ValueError:
    print("정수가 아닙니다.")
finally:
    print("입력 처리를 마쳤습니다.")
```

```python

while True:
    try:
        score = int(input("점수(0~100): "))
    except ValueError:
        print("숫자로 입력해 주세요.")
        continue
    if 0 <= score <= 100:
        break
    print("0부터 100 사이의 값을 입력해 주세요.")
print(f"입력된 점수: {score}")
```
