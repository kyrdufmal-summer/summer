## Part 1. 리스트와 반복문을 이용한 데이터 처리

점수 데이터를 리스트에 저장하고, `for`문과 `if`문을 사용해 80점 이상인 점수와 개수를 확인했습니다.

### 변수 초기화 위치 정리
- 전체 데이터를 계속 누적해서 계산할 값은 반복문 밖에서 초기화합니다.
- 한 사람, 한 주문, 한 파일처럼 각 항목마다 새로 계산할 값은 반복문 안에서 초기화합니다.
- 이번 과제의 `count`는 전체 점수 중 80점 이상인 개수를 누적해서 세는 값이므로 반복문 밖에서 `count = 0`으로 초기화했습니다.

```python
scores = [85, 72, 90, 65, 88]
count = 0

for score in scores:
    if score >= 80:
        count += 1
        print("80점 이상:", score)

print("80점 이상인 점수의 개수:", count)
```

## Part 2. 딕셔너리를 이용한 단어 빈도 분석

문장을 `split()`으로 단어 단위로 나눈 뒤, 빈 딕셔너리에 단어별 등장 횟수를 직접 저장했습니다.
처음 등장한 단어는 1로 저장하고, 이미 존재하는 단어는 기존 값에 1을 더했습니다.

```python
sentence = """
파이썬 공부는 재미있다
파이썬 공부는 어렵지만 재미있다
데이터 분석에도 파이썬을 사용한다
파이썬 공부를 계속하면 실력이 늘어난다
"""

words = sentence.split()
word_count = {}

for word in words:
    if word in word_count:
        word_count[word] += 1
    else:
        word_count[word] = 1

print("전체 단어 빈도")
for word, count in word_count.items():
    print(word, ":", count, "회")

print("\n2회 이상 등장한 단어")
for word, count in word_count.items():
    if count >= 2:
        print(word, ":", count, "회")
```

## Part 3. 함수로 반복되는 코드 정리하기

Part 2에서 작성한 단어 빈도 계산 코드를 함수로 분리했습니다.
문장을 함수에 전달하면 딕셔너리 형태의 계산 결과를 반환하도록 작성했습니다.

```python
def count_words(sentence):
    words = sentence.split()
    word_count = {}

    for word in words:
        if word in word_count:
            word_count[word] += 1
        else:
            word_count[word] = 1

    return word_count


def print_words_over_count(word_count, minimum_count):
    for word, count in word_count.items():
        if count >= minimum_count:
            print(word, ":", count, "회")


result = count_words(sentence)

print("함수 실행 결과")
for word, count in result.items():
    print(word, ":", count, "회")

print("\n2회 이상 등장한 단어")
print_words_over_count(result, 2)
```

아래에서는 사용자가 숫자가 아닌 값을 입력했을 때 발생하는 `ValueError`를 `try-except`로 처리했습니다.

```python
try:
    minimum_count = int(input("몇 회 이상 등장한 단어를 찾을까요? ") )
    print_words_over_count(result, minimum_count)
except ValueError:
    print("숫자를 입력해야 합니다.")
```

## 도전 문제 - 특정 단어 검색하기

사용자가 찾고 싶은 단어를 입력하면 해당 단어가 몇 번 등장했는지 출력하도록 작성했습니다.

```python
search_word = input("찾을 단어를 입력하세요: ")

if search_word in result:
    print(search_word, ":", result[search_word], "회")
else:
    print(search_word, "은(는) 문장에 없습니다.")
```
