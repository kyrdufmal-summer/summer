## Part 1. 리스트와 반복문을 이용한 데이터 처리

- 리스트 생성
- for 반복문으로 데이터 순회
- if 조건문으로 조건에 맞는 데이터 확인
- 조건을 만족하는 데이터 개수 계산
- 결과 출력

점수 데이터를 리스트에 저장하고,
for문으로 하나씩 확인하면서 80점 이상인 점수의 개수를 계산했다.
### count 변수 위치 정리

- 전체 데이터를 계속 누적해서 계산할 값은 반복문 밖에서 초기화한다.
- 한 사람, 한 주문, 한 파일처럼 각 항목마다 새로 계산할 값은 반복문 안에서 초기화한다.

이번 과제의 `count`는 80점 이상인 점수의 전체 개수를 누적해서 세는 값이므로 반복문 밖에서 `count = 0`으로 초기화했다.

```python
scores = [85,72,90,65,88]
count = 0

for score in scores:
    if score >= 80:
        #80점 이상인 점수의 개수 count 계산
        count += 1
        print ("80점 이상:",count,score)

```

## Part 2. 딕셔너리를 이용한 단어 빈도 분석

```python
sentence = """
파이썬 공부는 재미있다
파이썬 공부는 어렵지만 재미있다
데이터 분석에도 파이썬을 사용한다
파이썬 공부를 계속하면 실력이 늘어난다
"""

# 1. 문장을 단어 단위로 나누기
words = sentence.split()

# 2. 단어별 등장 횟수를 저장할 빈 딕셔너리 만들기
word_count = {}

# 3. 단어를 하나씩 반복하면서 등장 횟수 계산
for word in words:
    # 이미 나온 단어라면 기존 값에 1 증가
    if word in word_count:
        word_count[word] += 1

    # 처음 나온 단어라면 1로 저장
    else:
        word_count[word] = 1


# 4. 전체 단어와 등장 횟수 출력
print("전체 단어 등장 횟수")
print("-" * 30)

for word in word_count:
    print(word, ":", word_count[word], "회")


# 5. 2회 이상 등장한 단어만 별도로 출력
print()
print("2회 이상 등장한 단어")
print("-" * 30)

for word in word_count:
    if word_count[word] >= 2:
        print(word, ":", word_count[word], "회")
```

## Part 3. 함수로 반복되는 코드 정리하기

```python
# Part 3 코드는 여기서 작성합니다.

def count_words(sentence):
    # 문장을 단어 단위로 나누기
    words = sentence.split()

    # 단어 빈도를 저장할 빈 딕셔너리
    word_count = {}

    # 단어별 등장 횟수 계산
    for word in words:
        if word in word_count:
            word_count[word] += 1
        else:
            word_count[word] = 1

    # 계산 결과 반환
    return word_count


# 함수 실행
result = count_words(sentence)


# 전체 단어와 등장 횟수 출력
print("전체 단어 등장 횟수")
print("-" * 30)

for word in result:
    print(word, ":", result[word], "회")


# 특정 횟수 이상 등장한 단어 출력 함수
def print_words_over_count(word_count, minimum_count):
    print()
    print(minimum_count, "회 이상 등장한 단어")
    print("-" * 30)

    for word in word_count:
        if word_count[word] >= minimum_count:
            print(word, ":", word_count[word], "회")


# 2회 이상 등장한 단어 출력
print_words_over_count(result, 2)
```
